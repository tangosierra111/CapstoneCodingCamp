## 6.1. Data Cleaning - Missing Value Handling

```python
# Cek nilai yang kosong
df.isna().sum()
```

```python
# Cek value_counts() untuk fitur 'ai_tools_used'
df['ai_tools_used'].value_counts()
```

```python
# Cek value_counts() untuk fitur 'ai_used_purpose'
df['ai_usage_purpose'].value_counts()
```

```python
# Pada fitur 'ai_tools_used', alih-alih memberikan label,
# saya mengganti untuk setiap data yang bernilai 'ChatGPT+Gemini'
# dengan ChatGPT saja atau Gemini saja. Sehingga, konsekuensinya
# adalah kehilangan satu kolom tersebut.
mask = df['ai_tools_used'] == 'ChatGPT+Gemini'

# Mengambil indeks yang perlu dibagi
idx = np.random.permutation(df[mask].index)

# Shuffle
np.random.shuffle(idx)

# Bagi dua
half = len(idx) // 2

df.loc[idx[:half], 'ai_tools_used'] = 'ChatGPT'
df.loc[idx[half:], 'ai_tools_used'] = 'Gemini'
```

```python
# Mengatasi data yang hilang dengan strategi mode (most_frequent) untuk data kategorikal
df['ai_tools_used'] = df['ai_tools_used'].apply(lambda x: df['ai_tools_used'].mode()[0] if pd.isna(x) else x)
df['ai_usage_purpose'] = df['ai_usage_purpose'].apply(lambda x: df['ai_usage_purpose'].mode()[0] if pd.isna(x) else x)
```

```python
# Cek kembali jumlah data yang hilang
df.isna().sum()
```

## 6.2. Data Cleaning - Outlier Handling

```python
# Ambil setiap kolom numerik
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Calculate number of rows and columns for subplots
n_cols = 4  # You can adjust this number
n_rows = (len(num_cols) + n_cols - 1) // n_cols

# Create subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 4))
axes = axes.flatten()

# Cek outlier dengan visualisasi boxplot
for i, col in enumerate(num_cols):
    plt.sca(axes[i]) # Set the current Axes instance to the one at axes[i]
    plt.boxplot(df[col])
    plt.title(f'Boxplot of {col}')

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# Cek outlier dengan metode IQR
for col in num_cols:
    Q1 = df[col].quantile(.25)
    Q3 = df[col].quantile(.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(outliers[[col]])
```

```python
# Mengatasi data outlier dengan menggantinya (imputasi) dengan median
for col in num_cols:
    Q1 = df[col].quantile(.25)
    Q3 = df[col].quantile(.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].apply(lambda x: np.median(df[col]) if x < lower_bound or x > upper_bound else x)
```

## 6.5. Feature Engineering

```python
# Mengambil setiap kolom bertipe data string/object
cat_cols = df_cleaned.select_dtypes(include=['object', 'bool']).columns

for col in cat_cols:
    print(df[col].value_counts(), '\n')
```

```python
# Mengambil kolom dengan jenis data nominal kategorik
nom_cols = df_cleaned[['ai_tools_used', 'ai_usage_purpose']]

# Mengambil kolom dengan jenis data ordinal kategorik
ord_cols = df_cleaned[['grade_level', 'performance_category']]
```

```python
from sklearn.preprocessing import OneHotEncoder

# Label encoding untuk kolom dengan jenis data nominal dengan teknik One-Hot Encoding
encoder = OneHotEncoder(sparse_output=False)
encoded_nom = pd.DataFrame(
    encoder.fit_transform(nom_cols),
    columns=encoder.get_feature_names_out()
)

# Label encoding untuk kolom 'grade_level' dengan jenis data ordinal secara manual
ord_cols['grade_level'] = ord_cols['grade_level'].map({
    '10th': 0,
    '11th': 1,
    '12th': 2
})

# Label encoding untuk kolom 'performance_category' dengan jenis data ordinal secara manual
ord_cols['performance_category'] = ord_cols['performance_category'].map({
    'Low': 0,
    'Medium': 1,
    'High': 2
})
```

```python
# Menggabungkan seluruh kolom yang telah diterapkan label encoding
df_encoded = pd.concat([encoded_nom, ord_cols], axis=1)
```

```python
# Menggabungkan seluruh kolom awal dengan kolom yang telah diterapkan label encoding
df_final = pd.concat([df_cleaned.drop(cat_cols, axis=1), df_encoded], axis=1)
```

# 7. Model Building

## 7.1. Data Splitting

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import classification_report

# Split the features and the target variabel
X = df_final.drop('performance_category', axis=1)
y = df_final['performance_category']

# Split the features and the target into training set and testing set
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=1)
```

## 7.2. Model Development Using Scikit-learn

```python
# Initialize models
models = [
    LogisticRegression(),
    SVC(),
    DecisionTreeClassifier(),
    KNeighborsClassifier(),
    RandomForestClassifier(),
    GradientBoostingClassifier()
]

# Train each model
for model in models:
    # Fit the model
    model.fit(X_train, y_train)

    # Model prediction
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Training set evaluation
    train_report = classification_report(y_train_pred, y_train)

    # Testing set evaluation
    test_report = classification_report(y_test_pred, y_test)

    # Display the classification report for each model
    print('=' * 10 + ' ' + type(model).__name__ + ' ' + 10 * '=')
    print(f'Training Set Classification Report:\n{train_report}\n')
    print(f'Testing Set Classification Report:\n{test_report}')
    print()
```
